# Study 22 — Football from first principles

## Historical football: why clubs adopted common rules

**Living research notebook — updated 20 October 2026.** Detailed documentary evidence, quotations, and case histories are in [the Study 22 research log](../docs/research/22_football_club_origins.md), especially **“Working assessment — fairness, demonstrated skill, enjoyment, accessibility, adoption and spectators (20 October 2026)”**. This notebook records decisions and unresolved questions so that we do not repeatedly reopen arguments or manufacture a storyline.

**Central discipline:** Ask what people wanted, what alternatives they had, what contemporary sources tell us, and which conclusions are our plausible inferences. Don't assume standardisation was inevitable or beneficial in every respect.

## 1. Competitive fairness vs demonstrating skill

**Assessment: fairness is not an established driver; demonstration/recognition of skill is.**

- Two teams knowingly playing by one code can have a procedurally fair contest, even if one is more practised at it. Versatility is also a real sporting skill.
- One game under one code need not decide who is the better footballing side *across* codes. Teams could alternate codes or play returns. Common laws were only one possible response.
- **Contemporary evidence:** “Etonensis,” writing in October 1863 about Eton–Westminster compromise matches, complained that the rules inhibited players' preferred techniques and recognition beyond familiar circles; proposed delegates devise a universal game. The historical account is a *summary* of the letter, not yet checked original full text: https://playupliverpool.com/1863/10/06/lets-frame-rules-for-one-universal-game/
- **Possible inference:** a settled code lets participants specialise, demonstrate and compare developed skills against more clubs. The specialisation claim is our analysis, not a direct Etonensis quotation.

**Decision:** Do not investigate Victorian players' comparative athletic ability or invent exact performance handicaps. We have enough to explain what participants were questioning.

## 2. Enjoyment and preferences

- Rules about hacking, carrying, offside, fair catches and scoring determine the *kind of game* being played and which skills matter.
- **Contemporary evidence:** At the FA's December 1863 meeting, Blackheath's F. M. Campbell defended carrying/hacking as essential to the kind of game he valued; others argued the dangers discouraged participation. Blackheath withdrew when its preferred features were excluded: https://playupliverpool.com/1863/12/01/the-fourth-meeting-of-the-football-association/
- **Assessment:** different tastes help explain *resistance* to convergence. Popularity of the eventual winning code is not evidence it was inherently the most enjoyable.

## 3. Finding opponents — network effects, recruitment and competition

- **Poker analogy:** you might prefer a rare variant, but play Hold'em because that is where the players are. A club could value outside fixtures more than retaining every familiar rule.
- **Contemporary support, detailed in research log:** Louth/Lincoln coordination for fixtures even apart from FA membership (Tupholme, 1863); Richmond's intention to use FA laws without joining; Morley's expectation that enough clubs would give the common laws authority.
- **Hypotheses:** wider opponent networks, organised competitions (including FA Cup from 1871), institutional promotion, club recruiting and rule compromises could reinforce adoption.
- **Boundary:** “network effects” is an analytical label, not a quote from a Victorian. Having clubs increasingly use one code does not by itself identify each club's reasons or prove why *this* code spread.

**Decision:** evaluate competing causes against dated decisions rather than declaring a single victor.

## 4. Simplicity and learning the rules

- **Poker analogy:** Hold'em's basics are generally easier to teach than five-card pot-limit Omaha. Beginner accessibility is distinct from strategic depth and mastery.
- **Contemporary support:** John Charles Thring's 1861 article “Football, Simple and Universal” and his later *The Simplest Game* explicitly value simple, broadly usable laws: https://commons.wikimedia.org/wiki/File:Football,_Simple_and_Universal_(The_Field)_1861-12-28.png
- Contemporary FA consideration of Cambridge Rules reportedly praised simplicity; check the original meeting text before quoting: https://isfa.org.uk/our-history/the-history-of-isfa/
- **Assessment:** simplicity was considered when *designing* a common code; it is **not proved** to be why the association code *won*. Simplicity can aid explanation, recruitment, refereeing or agreement.

## 5. Spectators

- A spectator trained on one set of laws might have trouble understanding scoring, infringements, tactics or apparently surprising decisions in another code.
- Shared rules could make following matches across different clubs easier and possibly reinforce audience growth.
- **Counterargument:** code novelty can be entertaining, and early clubs did not necessarily prioritise paying spectators.
- **Status:** plausible modern inference; **contemporary spectators' arguments and actual causal importance not established**. Keep in notes; do not present as a documented FA motive.

## 6. Modern illustration — Clash of the Codes (1996)

- Two games: **Wigan beat Bath 82–6** under rugby league rules; **Bath beat Wigan 44–19** under rugby union rules.
- Both matches could be procedurally fair while measuring distinct, code-specific sporting abilities.
- Do not extrapolate score margins to nineteenth-century amateur players: radically different eras, skill specialisation and training environments.
- The separate **2015 Leicester Tigers vs American football converts** event was proposed but cancelled, not an actual crossover fixture.
- Context: https://www.superleague.co.uk/article/1392/hemmings-way-cross-coders

## 8. Who had authority? Public-school alumni, prestigious institutions and early FA formation

**New research correction (20 October 2026):** Earlier focus on Sheffield underplayed the distinct status and social networks of former public-school pupils; *but* avoid an equally false account of the FA being founded mainly by Eton/Harrow alumni.

**Primary voices and actions:**

- In November 1863, an Oxford undergraduate objected to the power of comparatively unfamiliar London clubs (Barnes, Blackheath) to 'dictate' rules to Eton, Harrow and Winchester, advocating leadership by the universities. This supports *one actor's legitimacy/status objection*, not general agreement. Locating historian: Graham Curry, https://ndl.ethernet.edu.et/bitstream/123456789/36949/1/GrahamCurry.pdf .
- In **1858–59**, letters from correspondents identifying with Rugby, Eton, Winchester and Marlborough disputed one another's codes, reportedly becoming abusive. A **pre-FA controversy** about schools' traditions and who should concede. Primary issue references and scholarly compilation: https://www.tandfonline.com/doi/full/10.1080/14660970.2022.2113128 ; FIFA Museum https://www.fifamuseum.com/fr/explore/fifamuseumplus/blog/The-quest-for-a-univ .
- In **November 1863**, Harrow's captain first asked what benefit FA membership gave and later refused; Charterhouse declined; Westminster deferred while a proposed public-school captains' meeting was pending. Contemporary report reproduced at https://playupliverpool.com/1863/11/10/the-football-association/ . Distinguish **school refusal** from alumni activity in adult clubs.
- At Cambridge, former pupils from schools with incompatible football traditions repeatedly developed compromise codes; the **1863 laws survive**, but parts of the 1840s history come from **1897 reminiscence**, not then-contemporary documents. https://specialcollections-blog.lib.cam.ac.uk/?p=5891 .

**Counterweight and correction:** Morley was not a major public-school alumnus. Research by Harvey and Aonuma challenges the claim that the 14 initial FA representatives were mostly alumni of the most prestigious public schools. https://www.jstage.jst.go.jp/article/jjpehss/59/2/59_13077/_article . More generally the founders of Sheffield FC, Creswick and Prest, were educated middle-class men, not an opposite 'working-class faction'.

**Old boys after 1863:** Old Harrovian Charles W. Alcock developed Forest/Wanderers, served on FA committee from 1866 and as FA secretary from 1870, and proposed the FA Cup in 1871 drawing on Harrow house competitions; **he did not attend the inaugural 1863 FA meeting**. https://www.thefa.com/news/2016/nov/02/history-of-the-fa-cup .

**Verdict:** Old-boy networks, school loyalties, elite status and universities need to feature as possible *mechanisms* of club organisation, persuasion and rule negotiation. At least some contemporaries explicitly viewed rule-making through status. **Do not claim prestige alone explains which rules ultimately prevailed.** Avoid exhaustive school-by-school biography. Full references and source-grading in [research log](../docs/research/22_football_club_origins.md), section “Incentive 5 addendum — ex-public-school networks, status, and legitimate authority”.

### Victorian status and class — essential context (20 October 2026)

**User's historical framing:** Do **not** read mid-Victorian arguments as if organisations were run primarily by today's commercial logic of 'who pays decides'. Public-school pedigree, university affiliation, gentlemanly respectability and social connections could carry authority independently of gate receipts or how many players preferred a code. Especially in **1863**, legitimacy could be claimed through *who one was* and *which institution one represented*, not only through the number of clubs signing up.

- Treat the Oxford undergraduate's complaint about Barnes/Blackheath 'dictating' to Eton/Harrow/Winchester as an example of **claimed social precedence**, not merely an efficiency objection.
- Distinguish **prestige to propose or legitimise rules** from **ability to organise games, persuade clubs or gain widespread acceptance**. A prestigious school might insist on its own game; a less prestigious club might have functioning fixtures and influence.
- Avoid anachronism: large spectator revenues and modern commercial leverage were **not** the starting conditions of the FA in 1863, although subscriptions, grounds, wealth, patronage and later gate receipts certainly mattered. Social class and money could reinforce one another; don't pretend economics was absent.
- The claim that prestige **decided the eventual standard code** remains an inference to test, not a demonstrated outcome. Do not replace a network-effect inevitability story with a Victorian-class inevitability story.

## 9. Working-class clubs might challenge old-school authority — but which decade and which issue?

**Hypothesis from discussion (20 October 2026):** Once working-class football teams became important, some may have questioned why socially prestigious Old Etonians or London gentlemen were entitled to determine football's rules and governance. A plausible class/status tension; don't attribute the exact words or psychology to unspecified Victorian workers without evidence.

### Chronology is decisive

- **1863 formation of FA:** organised association clubs were disproportionately middle-/upper-class, school-/university-/professional-linked. Do **not** project the later mass working-class football community backward into the inaugural FA meetings. Sheffield's founders were relatively prosperous educated men, not automatically a workers' rebellion.
- **1870s–1880s expansion:** industrial and northern clubs grew, giving them new sporting and organisational leverage. Old-boy dominance of the early FA Cup waned. Examples: Darwen vs Old Etonians (1879); Blackburn Rovers reached final against Old Etonians (1882); Blackburn Olympic, including industrial workers, beat Old Etonians in **1883**.
- **1884–85 challenge to governance:** northern clubs opposed the FA's anti-professionalism policy; a rival **British Football Association** was organised/threatened in **1884**; FA permitted professionalism in **1885**. **This is actual collective opposition to the governing body's policies**, though it is not evidence they were seeking to replace association's playing code or that each was explicitly attacking Etonian pedigree. FA institutional history: https://www.thefa.com/about-football-association/who-we-are/history .
- **Contemporary class framing:** reporting around the 1883 Blackburn Olympic–Old Etonians match is said to have described the opposition as 'patricians and plebeians'; a later account quotes papers describing Etonians as 'swells' and Blackburn players as men working for their bread. **Locate original dated newspaper before quoting as first-hand testimony**: https://gameofthepeople.com/2023/04/04/great-reputations-blackburn-rovers-1883-working-class-heroes/ . Academic study of the 1883 final cites contemporary press including 1882 Blackburn Standard fan letter: https://www.tandfonline.com/doi/full/10.1080/14660970.2016.1166771 . Note these quotes are *journalists' descriptions*, not players' confessions.

### What this does and does not show

- **Supported:** the FA's conception of gentlemanly amateur sport clashed with northern clubs' economic realities and aspirations; clubs actively forced a rules-of-governance compromise over payment. Working-class participation altered football's balance of influence.
- **Not proved:** an 1863 workers' movement protested Old Etonians' right to write the first FA code; the 1884 dispute was over *playing laws* rather than *professional status*; all northern clubs were working-class, or all southern clubs public-school aristocrats.
- **Interpretive angle:** Victorian status could be challenged in practice by participation, match results and organisational bargaining. Distinguish authority to set **laws of play** from regulation of **who may play and get paid**. This later development need not become the main story of the initial 1863 standardisation.


## 10. Scope correction — the question is “What is a game of football?”, not a history of football

**Editorial decision (20 October 2026):** Research into social class, rule-making authority, old boys, Sheffield, FA formation, and competing codes serves the central investigation of **what makes a game football**, rather than becoming a chronological survey of English football from 1863 to professionalism, league formation or modern commercialisation.

**Narrative destination:** What exactly were people deciding when they made football rules? What makes a given activity a game, what is its purpose, what abilities are rewarded, what behaviours are acceptable, when can people with different customs play together, and who may legitimately determine the answer? The 1863 debates are a historical case in **game definition, negotiated conventions, consent, competition and authority**.

**Scope discipline:**

- Sheffield matters as one functioning alternative tradition; avoid constructing an inevitable Sheffield-vs-FA showdown.

## 11. Authority — why did FA rules become the shared code? (20 October 2026)

**Question, not a football chronology:** Who gets to decide which agreed activities constitute 'football' when different groups have established legitimate games? How does *proposing* rules turn into having a *recognised authority* over a common game?

### Evidence-led steps

1. **1863: only voluntary authority.** The FA was a membership association, not a government empowered to dictate how schools, universities, Sheffield or other clubs played. In the _Bell's Life_ report of **14 Nov 1863**, E. C. Morley anticipated schools would retain their rules at first but see an advantage in a recognised code for cross-club matches; he considered 20–30 clubs sufficient to give it authority. This is **contemporary organiser intent/expectation**, not proof it was the sole cause: https://playupliverpool.com/1863/11/10/the-football-association/ .
2. **Weak early position and rival success.** Sheffield already had extensive club play and its own code; the early FA was fragile, possibly near dissolution in 1867. Historian Adrian Harvey (2001) attributes its survival partly to Sheffield's support. **Later historical interpretation, not primary testimony:** https://pubmed.ncbi.nlm.nih.gov/18578082/ . The claim that FA dominance was inevitable in 1863 is untenable.
3. **An incentive to join an organised shared activity.** In 1871 the FA established the **FA Cup**: only **15** clubs entered the first competition, from roughly **50** association members invited. Contemporary secretary Charles Alcock proposed a challenge cup open to FA members. The later institutional account preserves the proposal and figures: https://www.thefa.com/news/2016/nov/02/history-of-the-fa-cup . **Inference:** the competition made using FA rules rewarding beyond mere subscription to a code; don't claim without quantitative proof that it was the single decisive factor.
4. **Bargaining and mutual concessions, NOT simply London forcing Sheffield to surrender.** The FA incorporated some Sheffield features (e.g. the corner kick); differences narrowed over the 1870s. Sheffield pressed to retain its preferred touch restart while London wanted its offside. A Glasgow club, **Clydesdale**, proposed a compromise: throw-in in any direction. **17 Apr 1877** FA extraordinary meeting adopted the change **33–10**; **23 Apr** Sheffield association formally accepted the Clydesdale amendment and London rules. Contemporary press sources collated, with dates: https://nachspielzeiten.de/sheffield-fa-and-london-fa-1877-2/ ; **Sheffield and Rotherham Independent 24 Apr 1877** extract: https://www.gethistories.com/p/a-history-of-association-football-10e ; **Sheffield Independent 28 Apr 1877** called it an *amalgamation*: https://englandsoldestfootballclubs.com/140-years-ago-today-sheffield-association-and-the-football-association-amalagamate/ . **Caution:** those are modern sites reproducing contemporary records; inspect the actual original newspapers when publishing direct quotes.
5. **A recognised institutional arrangement followed.** County/district associations multiplied during **1875–85**, making the national association's laws more usable; later official FA history: https://www.thefa.com/about-football-association/who-we-are/history . Rugby maintained a separate governing tradition (RFU founded 1871), so the FA 'won' specifically within **association football**, not authority over every kind of football. Cross-nation law-making later passed to IFAB in **1886**: https://nationalfootballmuseum.com/items/the-laws-of-the-game-1863/ .

### Findings for the first-principles question

- **Who has authority?** Initially the participants/clubs who mutually consent to a game's rules; a standing association can become authoritative by coordinating a sufficiently useful community of players and matches.
- **What conferred wider acceptance?** Plausibly accumulating participants, organised competitive opportunity, publicity and administrative continuity **plus** negotiated compromises. The historical record supports each having some role but **does not measure relative causal weight**; do not turn 'network effects' into a single proven cause.
- **Victorian prestige mattered but was not self-enforcing.** Old-school authority competed with working clubs' competence and the FA's ability to organise. Historic actors differed on legitimacy. The 1877 result should not be narrated as an Etonian social-class victory.
- **What did the FA 'win'?** Recognition of its updated common playing laws for the amalgamated association code, rather than an inherent right to decree what all groups must call football.
- **Scope stop:** We now have a defensible answer for the **authority** argument. Do not expand into comprehensive football governance from 1880 onward or presume to solve the 'best' game. Subsequent discoveries may revise, not erase, this provisional conclusion.


## 12. The 1863 'coup' — Harvey's label tested against the meeting reports (20 October 2026)

**Research trigger:** Adrian Harvey, *An Epoch in the Annals of National Sport: Football in Sheffield and the Creation of Modern Soccer and Rugby* (2001), abstract: claims a **'coup'** by members opposed to Rugby-style football caused a split within weeks of FA formation. This is **Harvey's retrospective interpretation**, not a contemporary participant's own label. https://pubmed.ncbi.nlm.nih.gov/18578082/

### What contemporary evidence shows
- **26 Oct 1863:** Founding meeting attempted a common football code accommodating competing traditions.
- **17/24 Nov:** Draft by secretary E. C. Morley still provided (Rule IX) carrying after specified catches and (Rule X) tackling by charging/holding/tripping/hacking. On **24 November**, a division on *insisting on hacking in correspondence with Cambridge* reportedly passed **10–9**, showing the pro-hacking side could still win a close vote. **Don't mistake this narrow vote for final approval of all draft laws.**
- **Cambridge intervention:** Newly published Cambridge laws did **not** permit running with the ball or hacking. Morley brought them into the 24 Nov discussion, Alcock moved to treat them as the desirable basis; final settlement was delayed. A procedural dispute over the precise **wording and counting of the Cambridge motion** followed.
- **1 Dec contemporary transcript (_Bell's Life_, reported 12 Dec):** Blackheath's treasurer **Francis Maule Campbell** objected that at the previous meeting 7 backed his amendment ('worthy of consideration') and 8 backed Morley's wording praising Cambridge, but with **19 present** the president had not properly taken votes **against**. He said the resolutions should not have been entered/confirmed and formally protested. President **Arthur Pember** disputed his account, said a recount/second question had occurred and he had given a **casting vote**. **Do not present deliberate rigging as proved.** Direct reproduced 1863 proceedings: https://playupliverpool.com/1863/12/01/the-fourth-meeting-of-the-football-association/ ; shorter _Sporting Life_ version: https://en.wikisource.org/wiki/The_Sporting_Life/1863/Football_Association_Meeting .
- Campbell further accused the leadership of having postponed consideration because the 'hacking' advocates had been too strong at the previous meeting. Pember explicitly rejected the accusation as an allegation of ungentlemanly conduct; Morley defended delegates' right to propose a resolution; Alcock and Lawson denied premeditated obstruction.
- **1 Dec decisive changes:** Alcock proposed deleting Rules IX and X, Morley supported him. Campbell tried to **adjourn until school representatives could attend**: lost **4–13**. The motion to remove Rugby-style carrying and hacking **passed 11–5** (_Sporting Life_ contemporary transcript). Note the 13–4 tally was **against adjournment**, NOT the final 11–5 law decision.
- **8 Dec:** Blackheath, represented by Campbell, announced withdrawal over the new code; Rugby-style participants went their own way. RFU founded **1871**, not simultaneously with FA vote.
- **Important nuance:** FA's adopted 1863 laws did **not** ban all contact with the hands: catching after an opponent's touch (fair catch) still earned a mark and free kick; early laws also included try at goal. What was rejected was **running with the ball** and **hacking**, not every vestige of rugby-style football. https://nationalfootballmuseum.com/items/the-laws-of-the-game-1863/ and _Sporting Life_ text above.

### What counts as 'coup' and verdict
- **Established:** contentious reversal in political balance and outcome, postponement, change of attendance, disputed procedure/counting, formal protest, and withdrawal by a founding club.
- **Not established:** a secret prearranged conspiracy or objectively fraudulent ballot. Both sides publicly contested what was fair procedure.
- **Why it matters for “what is a game of football?”** A newly founded association tried to arbitrate between games whose core permitted actions **could not both be retained without changing the activity**. The decisive law-making result was **contingent on representation, procedure and consent**, not merely identifying the objectively best game. This supports our investigation of **who has authority to define the game**, while respecting historical uncertainty.

**Optional focused follow-up:** Retrieve exact full **24 November _Bell's Life_** scanned pages (the 1863 transcription web page currently displays them as nine images) to reconstruct disputed votes independently, plus 8 Dec _Sporting Life_ report to verify Blackheath's departure wording. The 1 Dec transcript already gives enough to explain what Harvey calls a coup; no need to chase tangential club biographies.


### Editorial relevance — decision after discussion

**Do not pursue the alleged 'coup' as an article storyline.** Factions, disputed counts, adjournments and vote reversals are ordinary organisational politics; sensational wording does not make them intrinsically revealing about the nature of football. Keep the contemporaneous record for historical accuracy, but don't go deeper into whether Pember/Morley manoeuvred against Campbell without a direct article need.

**Retain only what advances the main question:** (a) people favoured meaningfully different games (carrying/hacking versus restrictions), (b) the chosen laws defined which activities and skills counted as football *under the association code*, (c) objectors such as Blackheath could leave and continue with a different game. The *substance and consequences* of different game rules matter more than the internal mechanics of the vote.

**Conclusion:** Political manoeuvring is background, not an independent explanatory pillar. Return to **“What is a game of football?”**, not an FA factional history.


## 13. When does changing a rule make football a different game? (20 October 2026)

**Central question:** Is a football game defined by its exact laws, by the sort of contest they create, or by the people who agree to play it? Distinguish (i) a particular **match's agreed conditions**, (ii) a named **code/sport's continuous identity**, (iii) the wider **football family**. Don't presuppose a clean quantitative dividing line.

### Three concrete tests, with contrary evidence

**1. Fixed rules are not necessary for continuity of one named code.** The FA's original laws (1863) allowed a **fair catch earning a free kick**, gave a free kick at goal after an attacker first touched the ball behind the opposing goal line (Law 7), had **no crossbar**, and used a stringent offside restriction. Very different from modern association football, yet in the recognised institutional/history of the same code. **Source:** original 13 laws displayed and transcribed by National Football Museum, https://nationalfootballmuseum.com/items/the-laws-of-the-game-1863/ . Therefore *unchanged rules* cannot be a necessary condition for a sport's continuing identity, even though the precise laws are constitutive of any particular match.

**2. A change to the contest's tactics doesn't by itself produce a new sport.** Sheffield's original laws had no offside law, whereas the 1863 FA had a strict one. Sheffield FC's annual meeting **25 September 1865** resolved to play an offside rule but continue to use its old code with local clubs that did not adopt offside. Its original match notebook identified 1865 games under 'offside rules' versus 'old rules'. **Exceptional original archival documentation:** Sotheby's 2011 catalogue transcribing first-hand club minute books and match notes, https://www.sothebys.com/en/auctions/ecatalogue/2011/english-literature-history-childrens-books-illustrations-l11404/lot.10.html . Same club, same broad football tradition, two alternative rule arrangements for different matches. A strategically important change need not mean 'another sport'.

**3. Player consent enables a legitimate contest, but agreement alone cannot fix categorical identity.** Hypothetical: allow every football player to catch, carry or throw the ball. Agreement settles what is lawful *for this match* but not whether that match is association football, rugby, handball, or a bespoke football variant. Different communities may devise materially similar games independently. This is **analysis, not a documented 1863 quotation**.

### Distinctions and provisional finding

| What are we identifying? | Best basis | Limit |
| --- | --- | --- |
| What game is being played *in a particular fixture*? | Pre-agreed scoring objectives, actions, restrictions and their enforcement | Local/hybrid rules can differ by match |
| Is this the *same continuing sporting code*? | Historical and institutional continuity, recognised changes, participant conventions and organised competition | Not a precise mathematical threshold |
| Is it within the wider family of *football*? | Shared historical origins, conventions, broad resemblances and terminology | No single simple rule covers every code while excluding every other game |

**Most useful answer:** Rules define the *contest being played now*; a continuing community and institutional tradition determine whether successive rule-sets are treated as *the same evolving code*. The way rules organise skill and scoring matters substantively, but is neither a mechanical threshold nor the entire identity criterion. **'Football' can designate a family, while 'association football' identifies one evolving member.**

**Counterpoint worth preserving:** Don't say that naming alone makes any activity association football. Laws have practical constitutive consequences (allowed actions, goals, outcomes), and other participants/competitions may reject a private variation as outside their code. Likewise, historical continuity does not mean two codes share the same modern game when their agreed rules are incompatible.

**Editorial decision:** This is a direct consequence of our first-principles investigation, not a reason to launch a separate philosophical literature review. Once assessed, decide whether it materially sharpens the prospective article; avoid inventing a strict boundary that historical practice does not supply.


## 14. Source correction: Sheffield's early class composition, school links, and Cambridge's rule-making role (21 October 2026)

**User's corrective observation:** The label **'Sheffield'** evokes steelworks and working-class steelworkers to a modern reader, while the early club's founders were educated business and professional men who often carried school sport into adult life. Rebalance historical explanation towards **schools and universities as centres of developing and reconciling football codes**. Do not then assert that Eton or Cambridge owned or controlled football, or that all early Sheffield clubs comprised former elite public-school boys.

### Contemporary material and differently dated retrospectives

- **Sheffield FC founded 1857:** Nathaniel Creswick (solicitor, Sheffield Collegiate School) and William Prest (wine merchant, cricketer). Original 1858–59 membership lists, minutes and accounts survive. Sotheby's 2011 detailed archival catalogue identifies manufacturers, solicitors, merchants, doctors as common occupations; **this was not founded as a steelworkers' representative team**. https://www.sothebys.com/en/auctions/ecatalogue/2011/english-literature-history-childrens-books-illustrations-l11404/lot.10.html
- The same catalogue describes a club history written for its **1907 jubilee**, based partly on original members' reminiscences, explicitly stating that **former public-school men in professions and businesses** missed opportunities to play and **obtained rules from multiple public schools** when constructing their own. **Source status: 1907 retrospective explanation, not a verbatim 1857 founding minute; check if 'public school' is being used for minor/local schools as well as the top elite schools.** The 1858 draft-law minutes provide separately dated direct evidence that rules were deliberately negotiated and revised, but do not by themselves prove every member's schooling or which school codes were copied.
- **Important distinction:** Sheffield Collegiate School catered largely to middle-class pupils and was **not equivalent in prestige** to the major Clarendon 'public schools' such as Eton, Harrow or Rugby. Treat schooling categories precisely; don't classify all founders as elite Etonian/Harrovian old boys. Prest's brother Edward attended St John's College, Cambridge at the time of John Charles Thring, a possible conduit to wider school/university ideas. The link itself is documented via historian's archival catalogue; any transmission of rules is **speculative**, not proven.
- **Hallam FC founded 1860** from an established cricket club; John Charles Shaw, connected to its early development, was a clerk/law stationer. Hallam's own club history says it was designed to be **less exclusive** than Sheffield FC, which helps show distinctions between local clubs: https://hallamfc.co.uk/our-history/ . Don't infer all Sheffield-area clubs had identical membership or social character.
- **Historical counterweight:** Sheffield and other local communities nevertheless adapted their rules through their own experience, formed enduring playing institutions and sometimes resisted school-centred proposals; earlier regional and popular football outside elite schools also existed. School origins are a mechanism, **not exclusive authorship** of all football.

### Why school/university influence is a more natural *starting point* for rule authority

1. Different schools had **established games with particular laws, skills and traditions**. Players carried them into adulthood and valued that inherited legitimacy. That shaped *who was seen to have expertise*, but not necessarily a nationwide mandate.
2. **University attendance brought incompatible school traditions into daily contact.** Cambridge footballers therefore tried repeatedly to formulate a playable common code. The **1848 code does not survive**, and part of the account comes from an **1897 recollection**; the **c.1856 'Laws of the University Foot Ball Club'** survive in a Shrewsbury School copy; Cambridge University Library holds the **1863 revised rules**, used directly in FA discussion. https://specialcollections-blog.lib.cam.ac.uk/?p=5891 . Compare actual 1856 code https://en.wikisource.org/wiki/Cambridge_Rules_(1856) .
3. **Prestige vs decision rights:** schools and universities were natural sources of **social and intellectual authority** in early Victorian football, but Cambridge's code was principally for university participants. The schools maintained incompatible football traditions, rather than yielding to a single national school authority. The FA's 1863 meeting made a distinct **voluntary governing association for member clubs**, **borrowing Cambridge's rules** rather than commanding the university.
4. **London clubs were also socially connected to education:** FA's own history notes Morley was not an elite public-school old boy but Barnes included old boys from several schools; contested customs encouraged his proposal of common laws. https://www.thefa.com/about-football-association/who-we-are/history . Thus it is misleading to frame **Sheffield vs schools vs London** as three unrelated, socially homogeneous blocs.
5. University rule-making / school prestige **did not alone supply a nationwide organisation of consenting clubs and fixtures**. One code became institutionally authoritative through adoption by clubs, negotiated laws and shared competitions. Distinguish the **origin and perceived legitimacy of ideas** from their **wider adoption and continuing rule governance**.

### Editorial implication, bounded

**Revised causal pathway to test against evidence:** games and social norms developed in schools and other communities → alumni take traditions into university and adult leisure → university footballers face incompatible school codes and devise compromises → clubs such as Sheffield (educated professional founders) make *their own* practical adaptations → FA provides an institutional coordinating mechanism while adopting much of the Cambridge approach → continuing acceptance and negotiation gradually establishes the association code. **Not a single inevitable pipeline and not a claim every game's ancestry lies in elite schools.**

**Reader-orientation sentence worth considering:** 'Sheffield is famous for steel, but the men who founded Sheffield FC in 1857 were largely solicitors, manufacturers, merchants and doctors—not teams of steelworkers competing against old Etonians.' Check final wording against occupational sample: individual employees may have had other occupations and association evolved over time.

**Decision:** Strengthen school/university role in authority discussion. Preserve both distinction between local Collegiate and Clarendon elite schools and evidence gradings (contemporary record versus 1907 retrospective). Do **not** launch a comprehensive history of football; everything here serves **'what is a game of football, and who gets to decide?'**.


## 15. Inventing a sport vs governing an existing game — padel comparison (21 October 2026)

**First-principles distinction prompted by user:** A modern inventor can announce an original game's equipment, permitted actions and winning conditions; the FA in 1863 was **not the inventor of football**, and therefore could not naturally claim an original creator's authority. It sought agreement among communities already playing different established football games. Distinguish the **authority to propose one's invented rules** from **enduring authority over everyone who adopts or adapts a game**.

**Useful evidence against an over-simple contrast:** The International Padel Federation (FIP) records that **Enrique Corcuera developed padel in Acapulco in 1969**; his wife **Viviana wrote the first rules**. The game then spread through Spain and Argentina. **National organisations emerged** (Argentina 1988), the **FIP was founded in Madrid on 12 July 1991** by the Argentine, Spanish and Uruguayan associations, and **Spanish/Argentine rules were unified in 1997** after divergence. Primary institutional timeline: https://www.padelfip.com/history/ . So even a game with a traceable single inventor can *acquire variants and require negotiation* as participation spreads. A governing federation's rule-making legitimacy derives from member recognition and organisation, not simply an inventor saying 'I own the game'.

**Comparison to football:** Football's rules were already decentralised before the FA existed; padel began with a specified recognisable home-court design and written rules but decentralised as it spread. One is more visibly **codification of existing traditions**, the other begins as **a proposed invention**, but both face authority/coordination problems once people choose to play beyond the originator. Do not claim every new game is immune to disputes, or that copyright/patent or commercial ownership automatically gives authority over every independent informal game.

**Editorial status:** A compact modern contrast may clarify our central investigation. Do **not** launch a second history of padel; source finding is sufficient.


## 16. What did Hallam's 'less exclusive' claim actually mean? (21 October 2026)

**User's key question:** Earlier research repeated the modern Hallam FC website saying the football club was founded in 1860 to be **'less exclusive' than Sheffield FC**. Do not translate this into a proven working-class challenge unless admissions policies and member occupations substantiate it.

**Modern club account (not 1860 contemporary testimony):** Hallam FC history states that Hallam Cricket Club had over **300 members** by the 1850s and that cricket members familiar with Sheffield football wished to found a 'less exclusive club led by cricketers' in 1860: https://hallamfc.co.uk/our-history/ . **Unspecified:** less exclusive in which respect? Admission, social status, cost, number of potential players, geography, or distinct membership? Do not conflate the >300 **cricket** members with >300 founding Hallam **football** members.

**Strong direct evidence of Sheffield FC's exclusivity in procedure:** The Sheffield Football Club **1858 club rules**, transcribed at https://spartacus-educational.com/FsheffieldA.htm (section Primary Sources, Rules 5, 11, 12), stipulated a **2s 6d subscription**, nomination by one member and seconding by another, **committee ballot**, and exclusion by **two black balls**. A friend could be introduced only under conditions (Rule 15). Thus Sheffield was an **invitation/vetting-based members' club**, not free open-entry recreation. However the price itself should not be called prohibitively expensive without comparison to wages; procedural social gatekeeping is established independently of wealth.

**Important counterweight from Hallam founder:** In letter dated **18 October 1907**, John Charles Shaw, an original Sheffield FC member who later founded Hallam, said he persuaded **many Sheffield FC members to join Hallam** specifically 'for the purpose of getting a team to play against the old club' (Sotheby's catalogue of the original signed correspondence): https://www.sothebys.com/en/auctions/ecatalogue/2011/english-literature-history-childrens-books-illustrations-l11404/lot.10.html . His direct recollection **47 years after the events** demonstrates *creating an outside opponent* as one expressed motive and interlocking social membership, **not a documented class revolt**. No original Hallam football admission rules yet located to prove how membership restrictions actually compared.

**Decision:** Say **Hallam presents itself as less exclusive; Sheffield's original admission procedures show real selection; but exactly what Hallam meant and whether its own rules were more open remain unverified**. The old boys / educated middle-class distinction remains central. This example also demonstrates a different reason to found a second club: **a new external opponent**, despite partially overlapping membership.


## 17. How much did Sheffield FC membership cost relative to workers' wages? (21 October 2026)

**Price direct from 1857 Sheffield FC Rules and Regulations:** On admission, **2 shillings 6 pence** subscription, due again on **1 November annually** (Rule 5); committee could demand **a further call of up to 2s 6d within the year** (Rule 7); play days **Saturday 2 pm until dark** (Rule 10); candidate had to have proposer and seconder, with committee ballot and rejection by two black balls (Rules 11–12). The *primary rules* are transcribed at https://spartacus-educational.com/FsheffieldA.htm . Do not accept that site's introduction stating Creswick and Prest were two Harrovians; it conflicts with archival biographical evidence elsewhere; source here is **the document transcribed**, not all surrounding website commentary.

**Wages comparison — different sources and populations, not a single 'average Sheffield worker':**
- Board of Trade *Returns of Wages* (1887), Table Bb2 pp. 191–192, retrospective Sheffield cutlery series for **1860**, reproduced at https://history.pictures/2020/01/23/3-6-the-metal-industries-in-the-west-midlands/ : adult **table knife hafter 27s/week**; saw maker **32s**; razor forger **35s**; file forger **45s**; saw grinder **54s**. These are *specific skilled occupations* and are not typical wages for all workers or guaranteed week-to-week income.
- Ordinary agricultural labourer, England and Wales (selected farms) in 1860 **10s 11d/week**, from Board of Trade 1905 historical series quoted in 1911 Britannica: https://en.wikisource.org/wiki/1911_Encyclop%C3%A6dia_Britannica/Agriculture/Agricultural_Population_and_Wages . **Not a Sheffield wage**; included to show lower-wage contrast.
- Additional contemporaneous local adult evidence: a **1844** medical/statistical discussion of Sheffield silver-plating reports men earning **18s–42s**, women **8s–15s** a week, already illustrating considerable wage variation; https://pmc.ncbi.nlm.nih.gov/articles/PMC10367895/ . Different industry and 13–16 years earlier; do not apply as exact 1860 average.
- Local workers' schedule, **1867 Sheffield grinding wheels**, approximately 60 hrs weekly, operating **6 am–2 pm or 4 pm Saturdays**; some independent artisans regulated their own hours but work was still substantial. Historical thesis quoting 1867 inquiry: https://etheses.whiterose.ac.uk/5990/2/259670_VOL2.pdf . Source is a later thesis citing 1867 documents, *not direct newspaper testimony*. Access to Saturday football therefore varied by occupation/time arrangements.

**Computation:** 2s 6d = 30 pence (1/8 of £1) = 2.5 shillings.
- Annual subscription as proportion of **one week's** nominal skilled Sheffield pay: at 27s, **9.3%**; 32s, **7.8%**; 35s, **7.1%**; 45s, **5.6%**; 54s, **4.6%**. For a hypothetical 30s weekly wage, **8.3% of a week's gross pay**; at a six-day equivalent, ~half a day's wage. Dividing by weekly wage is **not** claiming that 5–9% of yearly pay was charged: the fee was *once per year*.
- At **10s 11d/week**, roughly **22.9% of one week's wages** or about 1.4 working days on a six-day assumption. Poorer family disposable income could be far tighter.
- If committee used Rule 7 to collect full extra call, **maximum stated annual subscription plus call** could be **5s**, excluding any personal kit, travel etc. **Do not claim the call was actually made** without accounts.

**Interpretation:** Money alone is not a strong explanation for Sheffield FC being socially exclusive: a *nominal* 2s 6d per year would be manageable for many skilled male artisans, though financially more difficult for low-paid workers or households under pressure. **Formal admissions depended on personal connections and committee approval**, and regular availability for **Saturday afternoon sport** may have mattered as much as or more than dues. Wages alone cannot prove how socially representative membership actually was. Keep distinction between **affordability** and **access**.

### Contradictory secondary descriptions — use original over confident generalisations

- Graham Curry (2023) and Harvey place **17 Sheffield Collegiate names among 57**. Kevin Neill challenges Shaw; cannot use these figures to establish *all other 40–41* schooling. **Confirmed attendance at one named institution** not same as *total school-educated*. https://www.cambridgescholars.com/resources/pdfs/978-1-5275-1277-1-sample.pdf printed pp.13–16. Especially no proof that ex-Collegians imposed their playing laws rather than negotiated or borrowed from another source.
- A different historical book, _Barbarians, Gentlemen and Players_ (1980s; online ANU PDF search-index) asserts **'Most members were old boys of Sheffield Collegiate School'** but gives no membership-based reconciliation and its page has major **names/date discrepancies** (e.g. spells Sorby Sorley; prints 1883 birth for a 1850s football committee member). https://openresearch-repository.anu.edu.au/bitstream/1885/115193/2/b12248885.pdf . Better to report **a sizeable minority positively identified (17/57)** and unspecified remainder than to endorse 'most' or claim zero unknown attended school.
- Another popular history claims **founder William Prest** attended Sheffield Collegiate: https://www.pitchpublishing.co.uk/sites/default/files/associated_downloads/Football%27s%2050%20Most%20Important%20Moments%20Sample.pdf ; school registration not demonstrated. Other websites inaccurately describe BOTH founders as Harrovians: example https://spartacus-educational.com/FsheffieldA.htm . Therefore celebrity/familiar-origin claims require individual vetting.
