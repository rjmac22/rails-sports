# Study 21 — Cricket from First Principles: Winning the Game

## What does it mean to win a game of cricket?

At the most basic level, cricket compares **runs**.

The two sides take turns trying to accumulate runs, and the result of the match ultimately depends on the relationship between the totals they produce.

This can be slightly obscured by the way cricket results are described.

A team might win:

- **by runs**;
- **by wickets**;
- or, in Test cricket, **by an innings and runs**.

But wickets are not a second kind of score.

If a team wins by six wickets, it has still won by passing the opposition's required run total. The six wickets describe how much batting capacity it had left when it did so.

So runs are the basic quantity being compared.

However, having more runs than the opposition at a particular moment is not always enough to win.

In Test cricket, for example, a side can be ahead on runs when the available time expires and the match can still be drawn.

So comparing run totals is fundamental to winning, but it is not the whole structure of the game.

## What opportunity does each team actually get to score its runs?

Cricket organises scoring into **innings**.

An innings is a period in which one team bats and attempts to accumulate runs while the other team bowls and fields.

The number of innings available depends on the form of cricket:

- in limited-overs cricket, each team normally has one innings;
- in Test cricket, each team can have two innings.

This means that a team's runs are not accumulated continuously throughout the whole match.

They are accumulated during particular opportunities to bat.

Those opportunities are themselves limited in some way. A batting side cannot simply continue an innings indefinitely and accumulate as many runs as it wants.

**Rules anchor:** MCC Law 13 defines the innings structure and allows matches to be played as one or two innings per side.

- MCC Law 13 — Innings: https://www.lords.org/mcc/the-laws/innings


## What limits a team's opportunity to keep scoring?

The answer is not quite the same in every form of cricket.

One important constraint appears throughout the game: **wickets**.

As wickets are lost, the batting side has fewer batters available. If ten wickets fall, an innings normally cannot continue.

But cricket also imposes another important constraint, and this differs by format.

In limited-overs cricket, the batting side is given a fixed allocation of deliveries.

In Test cricket, an innings has no fixed delivery allocation. Instead, the match itself is played within a limited amount of time.

So the two forms share part of the same structure, but not all of it.

It is therefore useful to follow them separately for a moment.

**Rules anchor:** MCC Law 13 provides the formal basis for innings ending through wickets, prescribed overs or prescribed time, depending on the match conditions.

- MCC Law 13 — Innings: https://www.lords.org/mcc/the-laws/innings


### How is a limited-overs innings constrained?

In limited-overs cricket, the batting side has two main finite resources:

- **wickets**;
- **deliveries**.

A team normally has eleven players. When a batter is dismissed, that batter can no longer contribute to the innings. Once ten wickets have fallen, there is normally only one batter left, so the innings cannot continue.

The batting side is also given a fixed allocation of overs.

For example:

- in T20 cricket, each team is normally allocated 20 overs;
- in a One Day International, each team is normally allocated 50 overs.

An innings can finish before either resource is fully used.

The batting side may lose all ten wickets before using all of its deliveries.

If it is batting second, it may also reach its target before using all of its wickets or deliveries.

So in limited-overs cricket, the opportunity to score is constrained by both **wickets remaining** and **deliveries remaining**.

**Rules anchor:** ICC format guidance distinguishes the standard delivery limits used in international limited-overs cricket: 20 overs in T20Is and 50 overs in ODIs.

- ICC — The Three Formats of Cricket: https://www.icc-cricket.com/about/cricket/game-formats/the-three-formats
- ICC — Playing Conditions: https://www.icc-cricket.com/about/cricket/rules-and-regulations/playing-conditions


### How is a Test innings constrained?

Test cricket has a different structure.

As in limited-overs cricket, wickets are finite. If ten wickets fall, the innings normally cannot continue.

But there is no fixed allocation of deliveries for a Test innings.

Instead, the **match as a whole is played within a scheduled time window**. Under current ICC men's Test playing conditions, a Test is scheduled for five days. The Home Board — the national board responsible for the home team and for staging the match — sets the daily hours of play before the match, subject to the ICC playing conditions, which provide for six scheduled hours of play per day.

Weather, bad light and other interruptions can reduce the amount of cricket actually played, although the playing conditions allow limited extensions and make-up time in specified circumstances.

The scheduled clock therefore sets the outer limit, but teams also have to judge how much play they are realistically likely to get before that limit. A forecast of heavy rain later in the day, for example, can reduce the expected amount of playable cricket before any time has actually been lost.

On the final day, the finish can become partly countable. Under the ICC cessation-of-play rules, the last hour normally requires a minimum of 15 overs, with play continuing until the relevant time and over requirements have been satisfied, subject to the playing-condition exceptions. Time remains the governing match constraint, but near the finish it can translate into a finite number of deliveries the batting side must survive.

An innings can therefore continue for as many deliveries as circumstances allow, but every period spent batting uses some of the finite playing opportunity available to complete the match. Tactically, teams are balancing both **scheduled time remaining** and **expected playable time remaining**.

A batting side can also choose to **declare** its innings closed before losing all ten wickets.

And when batting to a target, the innings can end because that target has been reached.

So in Test cricket, wickets constrain an individual innings directly, while **time constrains what can be achieved across the match as a whole**.

**Rules anchor:** ICC format guidance describes Tests as two-innings matches played over up to five days, without a fixed innings over limit. Clause 12 of the ICC Men's Test Match Playing Conditions governs the start and cessation of play, including the last-hour provisions.

- ICC — The Three Formats of Cricket: https://www.icc-cricket.com/about/cricket/game-formats/the-three-formats
- ICC — Playing Conditions: https://www.icc-cricket.com/about/cricket/rules-and-regulations/playing-conditions
- ICC — Men's Test Match Playing Conditions, Clause 12 (Start of Play; Cessation of Play)


## What do these two forms of cricket have in common?

The details differ, but both forms place limits on the opportunity to accumulate runs.

In both:

- runs are the quantity being accumulated;
- wickets restrict how long a batting side can continue to do so;
- the opportunity to keep scoring is limited in some additional way.

That additional limitation differs by format:

- in limited-overs cricket, the batting side has a finite number of deliveries;
- in Test cricket, the match has a finite amount of time.

So cricket is not simply a contest to score runs.

It is a contest to score runs **within constraints**.

Those constraints matter because they affect how many more runs a side may still be able to score and what it may still be able to achieve in the match.

## If runs determine who wins, why do wickets matter?

A wicket does not directly reduce the batting side's score.

Instead, it reduces the batting side's **future ability to score**.

When a batter is dismissed, that batter can no longer contribute to the innings. Another batter may replace them, but the number of available batters has been reduced.

Each wicket also moves the batting side closer to the point where the innings must end. Once ten wickets have fallen, the innings normally cannot continue.

That matters in both forms of cricket:

- in limited-overs cricket, losing wickets can prevent a side from making full use of its remaining deliveries;
- in Test cricket, losing wickets can end an innings even though substantial match time remains.

So wickets matter not because they are part of the score being compared, but because they reduce the batting side's **remaining scoring capacity**.

**Evidence anchor:** Headingley 2019 makes the mechanism visible. England were 286/9 chasing 359 when Jack Leach joined Ben Stokes. England still needed 73 runs, but one further wicket would have ended the innings.

## Is every wicket equally valuable?

No.

A wicket matters because it changes what the batting side can still achieve.

That means its value depends on the situation in which it falls.

Losing a wicket early in an innings may remove a batter who could otherwise have contributed for a long time.

Losing a wicket near the end of a limited-overs innings may matter less if only a few deliveries remain.

The identity and role of the batter matter too. Replacing an established specialist batter with a less accomplished batter may reduce the batting side's remaining scoring capacity, even though the incoming player is still a professional cricketer capable of surviving and contributing.

The number of wickets already lost can also matter. A wicket that brings the innings close to ending may have a much larger effect than one lost when many batters remain.

So a wicket does not have one fixed value.

Its importance depends on **how much it changes what the batting side can still do from the current situation**.

**Evidence anchor:** At Headingley in 2019, England's final wicket would have ended the chase immediately. Jack Leach was not a helpless batter — he was a specialist spin bowler who had made 92 as a nightwatchman earlier that summer — but Ben Stokes was already established on 61 and was far better placed to score the remaining 73 runs. That made preserving the final wicket and managing which batter faced the next delivery part of the same tactical problem.

## Is the value of the other constraints fixed either?

No.

Just as the value of a wicket depends on the situation, the importance of a remaining delivery or remaining time also depends on what can still be achieved with it.

In limited-overs cricket, a delivery early in an innings is one of many still available.

A delivery near the end may be one of only a few remaining opportunities to score.

In Test cricket, an hour early in the match is not necessarily equivalent to an hour late on the final day.

The value of that time depends on the score, the wickets remaining, which side is batting, and what result each side is trying to achieve.

Near the end of a Test, time and deliveries can also become linked. Once the last-hour provisions apply, a side may know that only a minimum number of overs or deliveries remain to be survived or used.

So the remaining opportunities in a cricket match do not have fixed values.

Their importance depends on the **current situation in the match**.

**Evidence anchor:** The reporting gives three contrasting examples in the same order used by the article. In Kolkata 2016, West Indies were running out of legal deliveries. In Sydney 2021, Australia were running out of time; in the final hour that remaining time became partly countable through the minimum-over requirement. At Headingley 2019, England were protecting their final wicket, and even a single run could be more or less valuable depending on whether it exposed Jack Leach or preserved Ben Stokes on strike.

## What do we mean by the state of the match?

The score alone does not tell us everything we need to know.

A team might be 120 runs for three wickets after 20 overs, or 120 for three with only five overs remaining.

The score is the same, but the situation is not.

To understand what the batting side can still achieve, we also need to know what opportunities and constraints remain.

In limited-overs cricket, relevant information includes:

- runs scored;
- wickets lost;
- deliveries remaining;
- whether the side is batting first or chasing a target.

In Test cricket, relevant information includes:

- runs scored by both sides;
- wickets lost;
- which innings is being played;
- which side is batting;
- the time remaining in the match.

Together, information like this describes the **match state**.

The match state is a compact description of the situation from which the next part of the game will unfold.

**Evidence anchor:** The match reconstructions show that score alone is insufficient to describe the situation. Relevant state information can include runs, wickets, target, deliveries or time remaining, and which batter is on strike. The exact components depend on the format and the problem being analysed.

- Reporting files: `world_t20_2016_final.md`, `sydney_2021.md`, `headingley_2019.md`, `control_of_next_delivery.md`


## Can we observe the changing match state in a market?

The match state is an analytical description. It is useful to ask whether people making decisions in real time also behaved as if those states had very different values.

Historical Betfair Exchange prices give one external observation of that process.

They are **not objective win probabilities**. An exchange price records where traders were willing to transact at that moment. But the prices are still useful evidence of how rapidly the possible results were being repriced as deliveries, time and wickets disappeared.

### Kolkata 2016 — deliveries

Betfair's Pro Match Odds data for market `1.123968779` shows the West Indies price moving approximately as follows during the final over:

| Match state | West Indies price | Approx. implied chance |
|---|---:|---:|
| 19 from 6 | 7.20 | 13.9% |
| 13 from 5 | 3.35 | 29.8% |
| 7 from 4 | 1.43 | 69.1% |
| 1 from 3 | 1.01 | 99.0% |

Three deliveries did not merely add eighteen runs. They transformed the market's assessment from West Indies as a large outsider to an almost certain winner.

### Sydney 2021 — time and changing objective

Betfair's Basic Match Odds data for market `1.177412195` is sampled at roughly one-minute intervals, so it is better suited to broad phases than individual deliveries.

At the start of the final day the approximate normalised split was Australia 84%, India 2.5%, draw 13%. During the Pant–Pujara partnership, India's price shortened to 3.6 — about a 28% implied chance. Around tea, after the chase had deteriorated, India's implied chance was back below 4% while the draw had risen to about 29%.

Around 30 overs remaining, the draw and an Australian win were almost level in the market. Later in the final session the draw moved above 80% and eventually towards certainty.

That movement mirrors the tactical change in the match. The relevant contest stops being primarily 'Can India reach 407?' and becomes 'Can Australia take five wickets before time disappears?'

### Headingley 2019 — the final wicket

Betfair's Basic Match Odds data for market `1.161492920` shows England trading around 30.0 after Stuart Broad's dismissal at 286/9, with Jack Leach arriving and 73 runs still required. Australia traded around 1.04.

That corresponds to roughly a 3.35% implied chance for England and 96.55% for Australia once the tiny draw probability is included.

For this article, one number is enough. It shows how strongly the market valued Australia's single remaining wicket relative to England's need for 73 runs.

### Methodological caution

The Kolkata archive is Pro tick-level data. Sydney and Headingley use Basic data sampled at roughly one-minute intervals. Live-blog timestamps are publication times rather than guaranteed ball-release times, so the Kolkata prices are representative observations from the stable intervals around the market step-changes, not claims about an exact millisecond price.

The market evidence supports the framework; it does not define it. The cricket rules and match events establish the state. The exchange prices show how strongly those changing states were valued by participants in real time.

## Does the batting side always want to score as many runs as possible?

No.

Runs are essential, but simply maximising the number of runs scored is not always the same as maximising the chance of winning.

In a limited-overs chase, the batting side only needs to reach the target. Once that happens, additional runs have no value because the match is already won.

Test cricket makes the distinction even clearer.

A batting side may be capable of continuing to score, but doing so consumes time that might instead be needed to dismiss the opposition. This is why a team may choose to declare while it still has wickets available.

A side batting last may also stop treating the target as its practical objective if the match state changes and preserving a draw becomes more valuable than continuing to chase a win.

So the batting side is not trying to maximise runs without regard to anything else.

It is trying to use the opportunities available to reach the most favourable match result.

**Rules anchor:** MCC Law 15 allows the captain of the batting side, other than in the final innings, to declare an innings closed while it is still in progress. That formal rule supports the point that maximising runs is not always the same as maximising the desired match result.

- MCC Law 15 — Declaration and Forfeiture: https://www.lords.org/mcc/the-laws/declaration-and-forfeiture

**Evidence anchor:** Sydney 2021 provides a concrete example of a batting objective changing with the match state. India began the final day with a target of 407 still available in principle, and Pant and Pujara briefly made victory look plausible. By the time Ravichandran Ashwin joined the injured Hanuma Vihari before tea, Pant and Pujara were out, Vihari had damaged his hamstring, Ashwin was struggling with his back and Ravindra Jadeja had a fractured thumb. Their practical problem became surviving roughly the next three hours for the draw. From that point, a dot ball could be valuable because it consumed Australia's remaining opportunity to take the wickets it needed.

## If runs, wickets, deliveries and time change in importance, how can we compare them?

They matter because they change what the two sides can still achieve in the match.

A run might move a chasing side closer to its target.

A wicket might make it much harder for that side to reach the target.

In a Test match, preserving wickets or consuming time might increase the chance of a draw, while scoring quickly might create more time in which to force a win.

So the useful common measure is not runs, wickets or time themselves.

It is their effect on the possible **result of the match**.

A match situation can therefore be thought of as more or less favourable depending on the prospects of winning, losing, drawing or tying from that position.

This gives us a way to understand why the same run or wicket can have very different importance in different situations.

**Evidence anchor:** The three reported matches support using the possible match result as a common way to compare otherwise different resources. In Kolkata, runs and balls remaining shaped whether West Indies could win before the legal deliveries ran out. In Sydney, wickets and remaining playing opportunity shaped whether India could secure a draw and whether Australia could still force a win. At Headingley, runs required, the final wicket and control of the next delivery all changed what results remained attainable.

This is an analytical framework rather than an official cricket definition or a numerical win-probability model.

- Reporting files: `world_t20_2016_final.md`, `sydney_2021.md`, `headingley_2019.md`


## How does the match move from one situation to another?

Cricket naturally proceeds through a sequence of deliveries.

Before each delivery, the match is in a particular state.

The delivery then produces an outcome and leaves the match in a new state.

That outcome might include:

- runs being scored;
- a wicket falling;
- no runs being scored;
- a delivery being used from a limited allocation.

So each ball gives us a discrete step in the progression of the match.

But the deliveries are not independent.

What happens on one ball changes the situation for the next.

A wicket may bring in a new batter. A single may change the striker. Runs may alter a target or required scoring rate. The fielding side may change its tactics.

Each delivery is therefore a separate event, but it takes place within a sequence in which earlier events affect what comes later.

**Rules anchor:** MCC Law 17 defines an over as six valid balls, giving cricket a formal delivery-by-delivery structure. No-balls, wides and certain dead-ball events do not count among those six valid balls.

- MCC Law 17 — The Over: https://www.lords.org/mcc/the-laws/the-over

**Evidence anchor:** The 2016 World T20 final gives a clean sequence of state changes: West Indies began the final over needing 19 runs from six legal balls; after successive sixes the problem became 13 from five, 7 from four, and 1 from three. Each delivery changed both the score requirement and the opportunities remaining.

- Reporting file: `reporting/cricket_winning_the_game/world_t20_2016_final.md`


## Can either side control what happens on the next delivery?

No.

Players make choices, but they cannot choose the outcome with certainty.

A batter can decide how aggressively to play.

A bowler can choose how and where to bowl.

The fielding side can decide where to position its fielders.

Those choices change which outcomes are more or less likely, but they do not determine what will happen.

A good tactical decision can still produce a bad result on one delivery, and a poor decision can sometimes produce a good one.

So tactical decisions are better understood as attempts to **change the probabilities of the possible outcomes**.

**Evidence anchor:** Headingley 2019 shows the difference between influence and control. England could refuse singles, seek twos and manage strike; Australia could change bowler and field placement. Yet Australia still created a dropped catch, a run-out chance and an lbw opportunity without converting any of them into the final wicket.

The language of changing the **probabilities** of possible outcomes is our analytical interpretation of those tactical choices, not a formal term from the Laws.

- Reporting files: `headingley_2019.md`, `control_of_next_delivery.md`


## What is the fielding side trying to achieve?

The fielding side is often described as trying to do two things:

- prevent runs;
- take wickets.

Both are important, but neither is the ultimate objective.

Preventing runs matters because it leaves the batting side in a less favourable position.

Taking a wicket matters because it reduces what the batting side can still achieve and may eventually end the innings.

Their importance depends on the match situation.

In one situation, preventing a boundary may be particularly valuable.

In another, accepting a greater risk of conceding runs may be worthwhile if it creates a better chance of taking a wicket.

So the fielding side is trying to influence the possible outcomes of play in ways that make the match situation more favourable to itself.

**Evidence anchor:** The fielding objective changes with match state. In Kolkata 2016, England did not need a wicket to win; they could win simply by using up the six legal deliveries without conceding 19 runs. In Sydney 2021, once India were primarily trying to survive, Australia needed wickets before the available time expired and crowded the bat in search of them. At Headingley 2019, Australia ultimately needed one wicket and later reflected that accepting greater boundary risk might have produced more deliveries at Jack Leach.

- Reporting files: `world_t20_2016_final.md`, `sydney_2021.md`, `headingley_2019.md`


## Can the fielding side maximise everything at once?

No.

The tactical objectives of the fielding side can conflict.

A choice that increases the chance of taking a wicket may also increase the chance of conceding runs.

A choice designed to prevent boundaries may reduce the opportunity to take a close catch.

A field that makes singles difficult in one area may leave another area more exposed.

So the fielding side faces trade-offs.

It has to decide which possible outcomes matter most in the current situation and how much risk it is prepared to accept in trying to produce them.

**Evidence anchor:** Headingley 2019 makes the trade-off visible. Australia spread the field to reduce some boundary risk against Ben Stokes, but that also made some singles and twos easier and reduced close-fielding options. Tim Paine later said there were moments when Australia might have been better accepting the risk of boundaries in order to get more deliveries at Jack Leach.

- Reporting file: `reporting/cricket_winning_the_game/headingley_2019.md`


## How can the fielding side influence those trade-offs?

The fielding side has several tactical choices.

The bowler can change how and where the ball is delivered.

The captain can change where the fielders are positioned.

These choices interact.

A bowling plan can make some shots or dismissals more likely than others, while the field can be arranged to respond to those possibilities.

Field placement is therefore not separate from the wider tactical problem.

It is one of the ways the fielding side changes which outcomes are easier or harder for the batting side to produce.

**Evidence anchor:** Headingley shows several interacting levers: field placement, bowler choice and bowling plan. Stokes said Nathan Lyon's return changed the tactical problem he faced, while Australia's field changes altered which scoring options were easier or harder to take.

- Reporting file: `reporting/cricket_winning_the_game/headingley_2019.md`


## Why can't the fielding side protect against every possible outcome?

Because the number of fielders is limited.

A fielder placed in one part of the field cannot simultaneously be somewhere else.

Strengthening one area therefore usually weakens another.

A fielder moved closer to the batter may create a better catching opportunity but leave more space further away.

A fielder moved towards the boundary may make it harder to score four runs in that area but make a single easier to take.

So field placement is a problem of allocating a limited number of players among competing tactical purposes.

**Rules anchor:** MCC Law 1 establishes that a side consists of eleven players. In ordinary play, once one player is bowling and one is acting as wicketkeeper, only nine players remain for the other fielding positions.

- MCC Law 1 — The Players: https://www.lords.org/mcc/the-laws/the-players

**Analytical inference:** Because those fielders are finite, placing one player in one area necessarily means that player cannot cover another area at the same time. Field placement is therefore a finite allocation problem. We do not need a separate authority to state that consequence; it follows directly from the limited number of available fielders.


## What does a field setting represent in this framework?

A field setting is one of the fielding side's tactical choices.

By placing its limited fielders in particular locations, the fielding side is choosing which possible outcomes it wants to make easier or harder.

Some positions may be intended primarily to create wicket opportunities.

Others may be intended to prevent boundaries, restrict singles or protect areas where the ball is especially likely to travel.

A single fielder can sometimes contribute to several of those purposes at once, but cannot cover every possibility equally.

So a field setting can be thought of as a **spatial allocation of limited resources in response to the probabilities and values of different possible outcomes**.

That gives us a more precise reason for wanting to describe where fielders stand.

Their positions are not merely traditional names on a cricket diagram. They are part of the mechanism through which the fielding side attempts to influence what happens next.

```python
# Sydney 2021: publication figure from Betfair Basic *full historical stream*
# Point SYDNEY_BETFAIR_ARCHIVE to the ORIGINAL market/event archive downloaded
# on 28 September 2026 (.tar.bz2, .bz2, .gz, or extracted JSON lines).
import bz2
import gzip
import hashlib
import json
import os
import tarfile
from pathlib import Path

import matplotlib.pyplot as plt
import matplotlib.dates as mdates
import pandas as pd

MARKET_ID = "1.177412195"
ARCHIVE = Path(os.environ.get("SYDNEY_BETFAIR_ARCHIVE", "")).expanduser()
if not os.environ.get("SYDNEY_BETFAIR_ARCHIVE") or not ARCHIVE.is_file():
    raise FileNotFoundError(
        "Find the original Betfair Basic archive downloaded on 28 September 2026, "
        "then set os.environ['SYDNEY_BETFAIR_ARCHIVE'] = '/full/path/to/archive' "
        "in a cell above this one. Six representative snapshots are NOT a "
        "substitute for the complete final-day archive."
    )


def iter_betfair_messages(path):
    """Read Betfair stream-format JSON lines from archives or extracted files."""
    if tarfile.is_tarfile(path):
        with tarfile.open(path, "r:*") as archive:
            for member in archive:
                if not member.isfile():
                    continue
                stream = archive.extractfile(member)
                if stream is None:
                    continue
                with stream:
                    if member.name.endswith(".bz2"):
                        source = bz2.BZ2File(stream)
                    elif member.name.endswith(".gz"):
                        source = gzip.GzipFile(fileobj=stream)
                    else:
                        source = stream
                    with source:
                        for line in source:
                            line = line.strip()
                            if line:
                                yield json.loads(line)
    else:
        opener = bz2.open if path.suffix == ".bz2" else gzip.open if path.suffix == ".gz" else open
        with opener(path, "rb") as stream:
            for line in stream:
                line = line.strip()
                if line:
                    yield json.loads(line)


def result_name(name):
    name = name.strip().lower()
    if name in ("draw", "the draw"):
        return "Draw"
    if name == "india":
        return "India"
    if name == "australia":
        return "Australia"
    return None


runner_names = {}
last_prices = {}
rows = []
segment = 0
status = None
in_play = False
updates_found = 0

for message in iter_betfair_messages(ARCHIVE):
    if message.get("op") != "mcm" or not message.get("pt"):
        continue
    timestamp = pd.to_datetime(message["pt"], unit="ms", utc=True)
    for change in message.get("mc", []):
        if str(change.get("id")) != MARKET_ID:
            continue
        updates_found += 1
        definition = change.get("marketDefinition") or {}
        if definition:
            status = definition.get("status", status)
            in_play = definition.get("inPlay", in_play)
            for runner in definition.get("runners", []):
                readable = result_name(runner.get("name", ""))
                if readable:
                    runner_names[runner["id"]] = readable
        if status != "OPEN" or not in_play:
            if last_prices:
                segment += 1  # do not connect lines across market suspensions
            last_prices.clear()
            continue
        if change.get("img"):
            last_prices.clear()  # replace the previous market image
        for runner in change.get("rc", []):
            if "ltp" in runner:
                value = runner["ltp"]
                if isinstance(value, (int, float)) and value > 1:
                    last_prices[runner["id"]] = float(value)
                else:
                    last_prices.pop(runner["id"], None)
        prices = {runner_names[k]: v for k, v in last_prices.items() if k in runner_names}
        if set(prices) != {"Australia", "India", "Draw"}:
            continue
        raw = {name: 1 / value for name, value in prices.items()}
        total = sum(raw.values())
        rows.append({
            "time_utc": timestamp, "segment": segment,
            **{name: 100 * raw[name] / total for name in raw},
        })

if not rows:
    raise ValueError(
        f"No complete in-play Australia/India/Draw snapshots for {MARKET_ID}. "
        f"Matching market messages: {updates_found}; runner mapping: {runner_names}. "
        "Confirm that you provided the original Sydney Match Odds archive."
    )

observed = pd.DataFrame(rows).sort_values("time_utc").drop_duplicates("time_utc", keep="last")
observed["time_sydney"] = observed["time_utc"].dt.tz_convert("Australia/Sydney")
observed = observed[
    observed["time_sydney"].dt.date == pd.Timestamp("2021-01-11").date()
].copy()
if observed.empty:
    raise ValueError("No observations during the Sydney-local final day (11 January 2021).")

# Plot the full observed series. Brief trading suspensions are routine in-play;
# splitting the line on every suspension made the figure look like confetti.
# Connect successive observed snapshots only across short gaps (<= 10 min).
# Do not carry prices forward, add synthetic points, or bridge longer gaps.
observed["plot_group"] = (
    observed["time_sydney"].diff().gt(pd.Timedelta(minutes=10)).cumsum()
)

fig, ax = plt.subplots(figsize=(11, 5.5), layout="constrained")
colours = {"Australia": "tab:orange", "India": "tab:blue", "Draw": "tab:green"}
for name in ("Australia", "India", "Draw"):
    for index, (_, frame) in enumerate(observed.groupby("plot_group", sort=True)):
        ax.plot(
            frame["time_sydney"], frame[name], linewidth=2.2,
            color=colours[name], label=name if index == 0 else None,
        )
ax.xaxis.set_major_locator(mdates.HourLocator(interval=1, tz=observed["time_sydney"].dt.tz))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%H:%M", tz=observed["time_sydney"].dt.tz))
ax.set(
    title="Sydney 2021: the draw becomes the contest",
    xlabel="11 January 2021 (Sydney local time, AEDT)",
    ylabel="Normalised market-implied share (%)",
    ylim=(0, 100),
)
ax.grid(alpha=0.25)
ax.legend(loc="center left", bbox_to_anchor=(1.01, 0.5), frameon=False)
fig.autofmt_xdate(rotation=0)
output_path = Path("../outputs/figures/sydney_2021_match_odds.png")
output_path.parent.mkdir(parents=True, exist_ok=True)
fig.savefig(output_path, dpi=220, bbox_inches="tight")
plt.show()

checksum = hashlib.sha256()
with ARCHIVE.open("rb") as source:
    for chunk in iter(lambda: source.read(8 * 1024 * 1024), b""):
        checksum.update(chunk)
print(f"Input: {ARCHIVE.resolve()}\nSHA256: {checksum.hexdigest()}\nMarket: {MARKET_ID}")
print(f"Source market messages: {updates_found}; plotted observations: {len(observed)}")
print(f"Local coverage: {observed.time_sydney.iloc[0]} to {observed.time_sydney.iloc[-1]}")
print(f"Exported locally: {output_path}")
print(
    "Method: Betfair Basic in-play last-traded prices; reciprocal odds normalised "
    "over three outcomes. Only valid in-play observations are plotted. "
    "Lines connect consecutive observed values across gaps up to 10 minutes, "
    "including brief suspensions; no intermediate values are asserted. "
    "Longer gaps are left blank. These are market-implied shares, NOT "
    "measured or objective win probabilities."
)
print("Sydney chart code version: v2 — fixed colours; continuous across short trading gaps")
```
