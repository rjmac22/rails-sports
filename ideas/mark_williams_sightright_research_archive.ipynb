## 8. Cumulative likelihood-ratio detector — method and all findings

**Fixed hypotheses:** H0 = baseline probability p0 for each opponent category; H1 = p0 + **0.06** probability points (chosen as a fixed improvement scenario, from earlier Studies 15–16). Six points is NOT a measured effect of coaching.

For a match with w winning and l losing frames:

    match_log_LR = w * log((p0+0.06)/p0) + l * log((1-p0-0.06)/(1-p0))
    cumulative_log_LR = sum(match_log_LR for ranked opponents to date)
    evidence_ratio = exp(cumulative_log_LR)

The old threshold was **LR ≥ 20:1**. These ratios concern two specific statistical hypotheses under strong assumptions; **they are not posterior odds of coaching success**, and the fixed-hypothesis formal optional-stopping result does NOT cover estimated p0, frame dependence and data-chosen groupings.

**Original 20:1 signals**:
- Two-year baseline: crossing after **353 usable ranked frames**, during German Masters **31 Jan–4 Feb 2018**, against Graeme Dott [26]. Earlier report: examined 77 ranked-opponent matches, 589 ranked frames, end-of-season LR **27.04:1**.
- Four-year baseline: crossing after **209 usable ranked frames**, during Northern Ireland Open **20–26 Nov 2017**, against Elliot Slessor [78].

**Intermediate diagnostic checkpoints**, where the left column is *all competitive frames elapsed* and the other columns cover only matches with known opponent ranks:

| Chronological total frames reached | Ranked frames used | Won | Expected won | Excess |
|---:|---:|---:|---:|---:|
| 50 | 50 | 35 | 31.2 | +3.8 |
| 100 | 100 | 70 | 61.6 | +8.4 |
| 150 | 143 | 96 | 87.5 | +8.5 |
| 250 | 226 | 149 | 135.2 | +13.8 |
| 350 | 321 | 207 | 189.8 | +17.2 |
| 500 | 482 | 298 | 277.4 | +20.6 |

**Frame-count trap:** detection after 209 or 353 **ranked-opponent frames** is not 209 or 353 total chronologically elapsed competitive frames. This distinction must survive editing.

**Ranking-group baseline used in two-year detector:** top16 182/404=0.4505; 17–32 133/220=0.6045; 33+ 132/209=0.6316.


```python
ref2 = pd.DataFrame([
    ("Top 16",182,404),("17–32",133,220),("33+",132,209)
], columns=["group","won","frames"])
ref2["p0"] = ref2.won/ref2.frames
ref2["p1"] = ref2.p0 + .06
ref2["log_per_win"] = np.log(ref2.p1/ref2.p0)
ref2["log_per_loss"] = np.log((1-ref2.p1)/(1-ref2.p0))
print(ref2.round(5).to_string(index=False))
print("Original LR boundary:",20,"; log boundary:",round(float(np.log(20)),4))
```

### Monte Carlo calibration — false alarms are not the same as formal p-values

Simulation preserved the actual **2017/18 opponent group and match-length sequence**, drew an underlying constant group ability per synthetic season from Beta(wins+0.5, losses+0.5), then drew match outcomes as Binomial(frames, group ability). Detector p0 remained the observed historical rate.

**Original two-year threshold:** **825 / 10,000 false signals = 8.25%**, worse than target **5%**.

For each synthetic season, record the **maximum cumulative log-LR**, then choose its **95th percentile** as the required detection threshold. This calibrates a **conditional synthetic-world** false-alarm probability, not a guaranteed real-world 5%.

**All calibrated post-coaching cumulative tests:**

| History | Historical ranked frames | Required LR | Williams's maximum LR | Improvement signalled? | First crossing |
|---|---:|---:|---:|---|---|
| **2 years (2015–17)** | **833** | **40.14:1** | **31.36:1** | **NO** | — |
| **3 years (2014–17)** | **1,212** | **29.64:1** | **47.31:1** | **YES** | **353 frames**, German Masters, Feb 2018 |
| **4 years (2013–17)** | **1,554** | **20.96:1** | **102.05:1** | **YES** | **209 frames**, Northern Ireland Open, Nov 2017 |

## 11. Modelling limitations and the selection issue the user spotted

**Knockout progression changes the opponent pool.** Winning an early match can lead to a much tougher later-round opponent. A player may become better yet have a falling raw frame percentage because their opponent mix changes. Rank stratification helps describe this, but it does not fix *selection into future opponents*.

Our simulations held the actual next opponents and match lengths fixed even in imagined seasons where Williams lost earlier and could not have reached later rounds. The simulations therefore **do not reproduce the sporting process generating the fixture path**.

Other constraints:
- No exact intervention date; unknown practice-frame changes.
- First ranking revision for a season may be dated after early matches.
- Missing rank coverage 22.3% before vs 4.8% after.
- Independent frames within matches are only an approximation; player form and opponents change.
- The three opponent buckets are crude proxies for difficulty.
- Fixed +6pp alternative is a design choice; not estimated coaching effect.
- Retrospective event crossing is **not a fully prospective rule**, since null calibration used the subsequently observed match schedule.
- Multiple baselines and methods were examined after results were seen.
- Binary frame results ignore richer performance measures (breaks, scoring patterns etc).

### Reproducing Monte Carlo detectors (only with match-level CSVs)

The helper below repeats the **same model family**: category-specific historical probabilities, a fixed **+0.06** alternative, Beta historical uncertainty, and the 95th percentile of maximum simulated detector evidence.

For each future match, it keeps the actual opponent category and number of frames while generating synthetic wins. It supports both **cumulative** and **resettable CUSUM** evidence.

**Not a scientific validation of that generative model**; only conditional numerical reproducibility. The reported figures above remain the source-of-record until a verified rerun matches. Random draws, seed, rank revision and match ordering can affect outputs.


```python
def replay_detector(history, future, baseline_seasons, *,
                    method="cumulative", seed=20261015, sims=10_000, shift=0.06):
    h=history[history.season.isin(baseline_seasons)].copy()
    h["group"]=group_by_rank(h["rank"])
    g=grouped(h)
    if g.won.isna().any() or (g.frames<=0).any():
        raise ValueError("Missing historical rank group")
    w0=g.won.to_numpy(dtype=int); n0=g.frames.to_numpy(dtype=int)
    rates=w0/n0
    f=future.copy()
    f["group"]=pd.Categorical(f.group,categories=groups)
    f=f.dropna(subset=["group"])
    idx=f.group.cat.codes.to_numpy()
    lengths=f.frames_played.to_numpy(dtype=int)
    p0=rates[idx]; p1=p0+shift
    lw=np.log(p1/p0); ll=np.log((1-p1)/(1-p0))
    delta=f.frames_won.to_numpy()*lw+f.frames_lost.to_numpy()*ll
    if method=="cumulative":
        observed=np.cumsum(delta)
    elif method=="cusum":
        cur=0.; tmp=[]
        for x in delta:
            cur=max(0.,cur+x);tmp.append(cur)
        observed=np.array(tmp)
    else:
        raise ValueError("method must be 'cumulative' or 'cusum'")
    rng=np.random.default_rng(seed)
    true_p=rng.beta(w0+.5,n0-w0+.5,size=(sims,3))
    running=np.zeros(sims);maximum=np.zeros(sims)
    for j,n in enumerate(lengths):
        sw=rng.binomial(int(n),true_p[:,idx[j]])
        d=sw*lw[j]+(n-sw)*ll[j]
        if method=="cusum":
            running=np.maximum(0.,running+d)
        else:
            running=running+d
        maximum=np.maximum(maximum,running)
    threshold=float(np.quantile(maximum,.95))
    hits=np.flatnonzero(observed>threshold)
    first=int(hits[0]) if len(hits) else None
    return {"method":method,"baseline_frames":int(n0.sum()),
            "analysed_frames":int(lengths.sum()),
            "threshold_score":threshold,
            "required_ratio":float(np.exp(threshold)),
            "max_score":float(observed.max()),
            "max_ratio":float(np.exp(observed.max())),
            "signal":first is not None,
            "first_ranked_frame":int(lengths[:first+1].sum()) if first is not None else None,
            "first_event":str(f.iloc[first].event) if first is not None else None,
            "simulated_false_alarm":float((maximum>threshold).mean())}

if pre_four_source is not None:
    for n in (2,3,4):
        years=["2013/2014","2014/2015","2015/2016","2016/2017"][-n:]
        print(n,"years",replay_detector(pre_four_source,replay_source,years))
    print("3-year CUSUM:",
          replay_detector(pre_four_source,replay_source,
              ["2014/2015","2015/2016","2016/2017"],
              method="cusum",seed=20261016))
else:
    print("Detector replay skipped: original match-level data unavailable.")
```

```python
# Optional pre-SightRight 2016/17 cumulative and resettable replay.
# Original extracted tables were REVERSE chronological within this subset.
# Independently verify the CSV's ordering before treating an event as first.
if pre_four_source is not None:
    watch=pre_four_source[
        pre_four_source.season.eq("2016/2017")].iloc[::-1].copy()
    watch["group"]=group_by_rank(watch["rank"])
    earlier=["2013/2014","2014/2015","2015/2016"]
    for method in ("cumulative","cusum"):
        print("2016/17",method,
              replay_detector(pre_four_source,watch,earlier,
                              method=method,seed=20261015))
else:
    print("Pre-SightRight replay skipped: match-level CSVs unavailable.")
```

We want **fairly simple research to tell an original story about real events**, not an editor needing to reproduce a methodological paper. Attempts to identify an exact improvement date led to changing baselines, Bayesian-flavoured null mixtures, likelihood ratios, CUSUM, unperformed power calculations and tournament-selection confounding. 2. Is a *specific sporting claim* about Williams's revival demonstrably wrong or incomplete? (A sports story must have a consequence, not only a metric.)
3. Can we compare the **same opponents or tournament rounds** without conditioning away the successful progression itself?
4. For a separate advanced article on **how hard it is to detect changing form**, how much power does the detector have when the true change is only six percentage points? (The proposed 10,000-season power cell has not been run.)

### Reusable lessons for the Rails Sports playbook
- Research **first**, story **after** findings; no reverse-engineering a desired conclusion.
- Ask early: **Where is the consequence?** Is the finding surprising and checkable?
- Inspect raw event-format semantics (Power Snooker points versus ordinary frames).
- Review missing records by opponent, not just a missing-data percentage.
- Separate frame percentages from underlying ability; schedule and opponent selection matter.
- Keep the denominator honest: total competitive frames versus usable ranked frames.
- Historical rank snapshots and retrospective calculations must not masquerade as genuinely real-time knowledge.
- Baseline variants are sensitivity checks, **not** independent confirming studies.
- A synthetic 5% threshold is **model-conditional**, not automatically an empirical 5% chance of error.
- Do not confuse no detection with proof of no improvement.
- Preserve rejected or null results alongside positive ones.
- Park work without deleting it; avoid endless research refinements when the editorial consequence isn't there.
